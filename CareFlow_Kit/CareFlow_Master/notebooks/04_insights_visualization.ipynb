# 04 - Insights, Bottlenecks & Visualisation

```
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv('../data/processed/event_log_clean.csv')
ca = pd.read_csv('../data/raw/case_attributes.csv')
```

```
bt = df.groupby('activity').agg(events=('activity','size'), avg_wait=('wait_minutes','mean'),
        avg_service=('service_minutes','mean')).sort_values('avg_wait', ascending=False)
bt.round(1)
```

```
bt['avg_wait'].plot(kind='barh', figsize=(8,5), title='Bottlenecks - average wait (min)')
plt.gca().invert_yaxis(); plt.tight_layout(); plt.show()
```

```
dep = df.groupby('department').agg(cost=('cost_inr','sum'), events=('activity','size')).sort_values('cost', ascending=False)
dep['cost'].plot(kind='bar', figsize=(8,4), title='Cost by department (INR)'); plt.tight_layout(); plt.show()
```

```
ca['bucket'] = pd.cut(ca.los_hours, [0,6,12,18,24,999], labels=['<6h','6-12h','12-18h','18-24h','>24h'])
ca.groupby('bucket', observed=True).agg(cases=('case_id','size'), avg_cost=('total_cost_inr','mean')).round(0)
```
