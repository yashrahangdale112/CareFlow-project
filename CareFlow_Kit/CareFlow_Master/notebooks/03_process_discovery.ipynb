# 03 - Process Discovery
Directly-Follows Graph + variants. pm4py mile to Inductive Miner bhi.

```
import pandas as pd, networkx as nx, matplotlib.pyplot as plt
df = pd.read_csv('../data/processed/event_log_clean.csv', parse_dates=['timestamp'])
```

```
df['next_activity'] = df.groupby('case_id')['activity'].shift(-1)
edges = df.dropna(subset=['next_activity']).groupby(['activity','next_activity']).size().reset_index(name='count')
edges.sort_values('count', ascending=False).head(15)
```

```
G = nx.DiGraph()
for _,r in edges.iterrows(): G.add_edge(r.activity, r.next_activity, weight=r['count'])
nx.draw(G, with_labels=True, node_size=1800, font_size=7, arrowsize=12)
plt.tight_layout(); plt.show()
```

```
df['variant'] = df.groupby('case_id')['activity'].transform(lambda s: ' -> '.join(s))
var = df.groupby('case_id')['variant'].first().value_counts().reset_index()
var.columns=['variant','cases']; var['pct'] = (100*var.cases/var.cases.sum()).round(2)
print('variants:', len(var)); var.head(10)
```
